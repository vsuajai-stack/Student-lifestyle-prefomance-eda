```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

# Ensure dataset is loaded and cleaned appropriately
df = pd.read_csv('student_lifestyle_dataset.csv')
time_cols = ['Physical_Activity_Hours_Per_Day', 'Extracurricular_Hours_Per_Day', 'Study_Hours_Per_Day','Sleep_Hours_Per_Day','Social_Hours_Per_Day']
df['total_hours'] = df[time_cols].sum(axis=1)
impossible_mask = df['total_hours'] > 24
for col in time_cols:
    df.loc[impossible_mask, col] = (df.loc[impossible_mask, col] / df.loc[impossible_mask, 'total_hours']) * 24
df['total_hours'] = df[time_cols].sum(axis=1)

sns.set_theme(style="whitegrid")

# Descriptive Statistics
print("=== Descriptive Statistics for Numerical Columns ===")
display(df.describe())

# Categorical counts
print("\n=== Stress Level Frequencies ===")
display(df['Stress_Level'].value_counts())

# Visualizations
fig, axes = plt.subplots(2, 3, figsize=(18, 10))
fig.suptitle('Distribution of Student Lifestyle Metrics and Performance', fontsize=16, fontweight='bold')

for i, col in enumerate(time_cols + ['GPA']):
    ax = axes[i//3, i%3]
    sns.histplot(df[col], kde=True, ax=ax, color='skyblue')
    ax.set_title(f'Distribution of {col.replace("_", " ")}')
    ax.set_xlabel('')

plt.tight_layout()
plt.show()
```

**Bivariate Analysis**

Bivariate analysis allows us to discover relationships and patterns between pairs of variables. Here we analyze:
1. **Numerical-to-Numerical**: Study Hours per day vs. GPA using a scatter plot and a linear regression line to assess academic yield.
2. **Numerical-to-Categorical**: GPA distribution grouped across different Stress Levels using box plots to evaluate stress-associated performance shifts.

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
df=pd.read_csv('student_lifestyle_dataset.csv')
# Relationship : GPA vs Stress Level
plt.figure(figsize=(10, 5))
sns.boxplot(data=df, x='Stress_Level', y='GPA', order=['Low', 'Moderate', 'High'], palette='muted')
plt.title('Academic GPA Distribution Across Stress Levels', fontsize=14, fontweight='bold')
plt.xlabel('Stress Level')
plt.ylabel('GPA')
plt.show()
```

**Multivariate Analysis**

Multivariate Analysis examines three or more variables simultaneously to grasp complex systems.
- We generate a complete **Correlation Heatmap** to explore how all numeric habits relate to one another and GPA.
- We display a **Grouped Scatter Plot** examining GPA versus Study Hours, colored dynamically by Stress Level to see if academic devotion affects stress distribution.

```python
# 1. Correlation Heatmap
plt.figure(figsize=(10, 7))
correlation = df.drop(columns=['Student_ID', 'total_hours', 'Stress_Level']).corr()
sns.heatmap(correlation, annot=True, cmap='coolwarm', fmt='.2f', linewidths=0.5)
plt.title('Multivariate Correlation Heatmap', fontsize=14, fontweight='bold')
plt.show()

# 2. Grouped Scatter Plot
plt.figure(figsize=(10, 6))
sns.scatterplot(data=df, x='Study_Hours_Per_Day', y='GPA', hue='Stress_Level', palette='Set1', alpha=0.7)
plt.title('GPA vs. Study Hours Segmented by Stress Level', fontsize=14, fontweight='bold')
plt.xlabel('Study Hours Per Day')
plt.ylabel('GPA')
plt.show()
```

**Outlier Analysis**

Outliers can dramatically skew statistical metrics and models. Here, we identify any statistical anomalies in the key output metric, **GPA**, using the Interquartile Range (IQR) method and decide on their handling strategy.

```python
# Calculate Interquartile Range for GPA
Q1 = df['GPA'].quantile(0.25)
Q3 = df['GPA'].quantile(0.75)
IQR = Q3 - Q1
lower_limit = Q1 - 1.5 * IQR
upper_limit = Q3 + 1.5 * IQR

outliers = df[(df['GPA'] < lower_limit) | (df['GPA'] > upper_limit)]
print(f"Lower IQR Limit for GPA: {lower_limit:.3f}")
print(f"Upper IQR Limit for GPA: {upper_limit:.3f}")
print(f"Total Outliers Identified: {len(outliers)}")
display(outliers)

# Boxplot for Outliers visualization
plt.figure(figsize=(8, 4))
sns.boxplot(x=df['GPA'], color='salmon')
plt.title('Identifying GPA Outliers with Boxplot representation', fontsize=14, fontweight='bold')
plt.xlabel('GPA')
plt.show()
```
